**What to do:** Choose an open dataset and a target variable to predict. Load it and report its basic shape and structure.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.set_printoptions(precision=4, suppress=True)

df_raw = pd.read_csv("pesodollar.csv", header=None, skiprows=7)

df = df_raw.iloc[:, [1, 2, 3]].copy()
df.columns = ["Year", "Month", "Average"]

df["Year"] = df["Year"].ffill()

df = df.dropna(subset=["Average", "Month"])
df["Average"] = pd.to_numeric(df["Average"], errors="coerce")
df = df.dropna(subset=["Average"]) 

df["Date"] = pd.to_datetime(df["Month"].str.strip() + " " + \
                            df["Year"].astype(str).str.replace(r'\.0$', '', regex=True), 
                            format="%B %Y", errors="coerce")
df = df.dropna(subset=["Date"]).sort_values("Date").reset_index(drop=True)

df = df[df["Date"] >= "1990-01-01"].reset_index(drop=True)

df["CurrentRate"] = df["Average"]
df["Trailing3MonthAverage"] = df["Average"].rolling(window=3).mean()
df["YearOverYearChange"] = df["Average"] - df["Average"].shift(12)
df["NextMonthRate"] = df["Average"].shift(-1)

feature_names = ["CurrentRate", "Trailing3MonthAverage", "YearOverYearChange"]
df_clean = df.dropna(subset=feature_names + ["NextMonthRate"])

X_raw = df_clean[feature_names].to_numpy()
y = df_clean["NextMonthRate"].to_numpy()

assert X_raw is not None and y is not None, "Load your dataset above before continuing."
n_samples, n_features = X_raw.shape
print(f"n_samples = {n_samples}, n_features = {n_features}")
print(f"feature_names = {feature_names}")
```
